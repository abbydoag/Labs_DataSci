## Origen de los datos

Se usa el dataset **ASL Alphabet** de Kaggle, `grassknoted/asl-alphabet`. Son
fotografias reales de 200x200 pixeles a color organizadas en carpetas por clase,
con 29 clases en total, las 26 letras mas SPACE, DELETE y NOTHING.

El dataset completo pesa cerca de 1 GB, por lo que **no se versiona ni se guarda
dentro del repositorio**. Todo el acceso se hace con la API de Kaggle en dos
modos distintos, y ninguno descarga el archivo completo:

1. **Inventario sin descarga.** `list_dataset_files` devuelve el listado de las
   87,000 rutas con su tamano en bytes. Con eso se resuelve la distribucion de
   clases y el balance del dataset sin bajar una sola imagen.
2. **Descarga dirigida por archivo.** `kagglehub.dataset_download` acepta el
   argumento `path`, que trae un archivo individual. Solo se descargan las
   imagenes de la submuestra que realmente se usa.

Las imagenes que si se descargan quedan en la cache de kagglehub, en
`~/.cache/kagglehub/datasets/grassknoted/asl-alphabet/`, fuera del repositorio.
La carpeta `Lab3/data/` esta en el `.gitignore` salvo `team-dataset/`, que
guarda las fotos propias del ejercicio 8.

## Requisitos

El laboratorio trae su propio entorno virtual con Python 3.13.5. Desde la
carpeta `Lab3` se crea y se instala asi:

```
python3 -m venv .venv
.venv/bin/python -m pip install -r requirements.txt
```

`requirements.txt` fija las nueve dependencias que los notebooks importan de
verdad. En Jupyter o VS Code hay que seleccionar el interprete `.venv` antes de
correr las celdas. El entorno no se versiona, esta en el `.gitignore`.

La API de Kaggle necesita un token propio. Se genera en kaggle.com/settings, en
la seccion API, y se guarda en `~/.kaggle/access_token` con permisos 600. Sin
ese archivo las celdas de descarga fallan con un error de credenciales.

Los modelos se construyen con PyTorch. El backend MPS esta disponible en
equipos Apple con chip propio y se usa para acelerar el entrenamiento.
